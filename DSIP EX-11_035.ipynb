```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread(r"C:\Users\Dell\Downloads\test.png", 0) # Load as grayscale
```

```python
# 1. Create a rectangular structuring element (5x5)
kernel_rect = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

# 2. Create an elliptical structuring element (5x5)
kernel_ellipse = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))

# 3. Create a cross-shaped structuring element (5x5)
kernel_cross = cv2.getStructuringElement(cv2.MORPH_CROSS, (5, 5))
```

```python
kernel = np.ones((5, 5), np.uint8)
```

```python
erosion = cv2.erode(image, kernel, iterations=1)

dilation = cv2.dilate(image, kernel, iterations=1)

opening = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel)

closing = cv2.morphologyEx(image, cv2.MORPH_CLOSE, kernel)

gradient = cv2.morphologyEx(image, cv2.MORPH_GRADIENT, kernel)

tophat = cv2.morphologyEx(image, cv2.MORPH_TOPHAT, kernel)

blackhat = cv2.morphologyEx(image, cv2.MORPH_BLACKHAT, kernel)
```

```python
plt.figure(figsize=(12, 8))
plt.subplot(2, 3, 1), plt.imshow(image, cmap='gray'), plt.title('Original Image'), plt.axis('off')
plt.subplot(2, 3, 2), plt.imshow(erosion, cmap='gray'), plt.title('Erosion'), plt.axis('off')
plt.subplot(2, 3, 3), plt.imshow(dilation, cmap='gray'), plt.title('Dilation'), plt.axis('off')
plt.subplot(2, 3, 4), plt.imshow(opening, cmap='gray'), plt.title('Opening'), plt.axis('off')
plt.subplot(2, 3, 5), plt.imshow(closing, cmap='gray'), plt.title('Closing'), plt.axis('off')  
plt.subplot(2, 3, 6), plt.imshow(gradient, cmap='gray'), plt.title('Gradient'), plt.axis('off') 
```
